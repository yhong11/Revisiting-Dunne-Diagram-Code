```python
regime = pd.DataFrame(index=range(len(gauge_id)),columns=['domination','id'])
for ii in range(len(gauge_id)):
       stationid=gauge_id[ii]
       stationid = float(stationid)
       camels_info['gauge_id'] = camels_info['gauge_id'].astype(float)
       aridity = camels_info[camels_info['gauge_id'] == stationid]['aridity'].iloc[0]
       aridity = round(aridity,3)

       stationid = "%08d"%stationid
       #6 kge
       outd4r = pd.read_csv('D:/yuanqi/new_pdm/hourly_sim/'+ str(stationid) + '_AI_' + str(aridity) + '_simulated26.csv',index_col=0)
      
       outd4 = outd4r[730:].sum()
       dqsim4 = outd4['Qsim']

       ri = outd4['ri']
       rs_all = outd4['rs']
       rg = outd4['rg']
       rd = outd4['rd']
       qd = outd4['qd']
       qb = outd4['qb']
       rs = np.array(rs_all-rg)


       if qb - qd > 0:
              regime['domination'].iloc[ii] = 'Baseflow'

       elif abs(ri/rd-rs/rd) < 0.1: 
              regime['domination'].iloc[ii] = 'Saturation excess & Infiltration excess'
       elif rs - ri> 0:
              regime['domination'].iloc[ii] = 'Saturation excess'
       elif ri - rs > 0:   
              regime['domination'].iloc[ii] = 'Infiltration excess'
       else:
              regime['domination'].iloc[ii] = 'DONTKNOW'

       regime['id'].iloc[ii] = stationid
regime.to_csv('D:/yuanqi/new urania partition/paper_figures/domination.csv')
```
