```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.cluster.hierarchy import linkage, fcluster
from sklearn.preprocessing import StandardScaler

# =========================
# Read data
# =========================
full_df = pd.read_csv(
    '/Users/yuanqihong/Desktop/URANIA_partitioning/camels_600_merged_all.csv'
)

component_ratio = pd.read_csv(
    '/Users/yuanqihong/Desktop/URANIA_partitioning/longterm_dunne_data.csv'
)

full_df = full_df.merge(
    component_ratio[['gauge_id', 'ri_ratio', 'rs_ratio', 'qb_ratio']],
    on='gauge_id',
    how='left'
)

# =========================
# Features used for clustering
# =========================
features = [
    'aridity',
    'slope_mean',
    'soil_depth_statsgo',
    'soil_porosity',
    'soil_conductivity',
    'frac_forest'
]

# =========================
# Hierarchical clustering
# =========================
scaler_cluster = StandardScaler()
data_scaled = scaler_cluster.fit_transform(full_df[features])

Z = linkage(data_scaled, method='ward')


# =========================
# Dunne coordinate function
# =========================
def compute_dunne_coords(means):
    # x-axis: wetter + more forested -> larger x
    x = -means['aridity'] + means['frac_forest']

    # y-axis: gentler slope + shallower soil + lower porosity -> larger y
    y = (
        -means['slope_mean']
        -means['soil_depth_statsgo']
        -means['soil_conductivity']
    )

    return x, y


# =========================
# Rescale coordinates to plotting space
# =========================
def normalize_coords(x_vals, y_vals, margin=0.15):
    x_arr = np.array(x_vals)
    y_arr = np.array(y_vals)

    x_norm = (x_arr - x_arr.min()) / (x_arr.max() - x_arr.min())
    y_norm = (y_arr - y_arr.min()) / (y_arr.max() - y_arr.min())

    x_scaled = margin + (1 - 2 * margin) * x_norm
    y_scaled = margin + (1 - 2 * margin) * y_norm

    return x_scaled, y_scaled


# =========================
# Plot function
# =========================
def plot_dunne_layout(group_means, group_dominants, n_clusters, group_size):
    x_raws, y_raws = zip(*[compute_dunne_coords(m) for m in group_means])
    x_pos, y_pos = normalize_coords(x_raws, y_raws)

    fig, ax = plt.subplots(figsize=(9, 6))

    fig.patch.set_facecolor("white")
    ax.set_facecolor("white")

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xticks([])
    ax.set_yticks([])

    ax.annotate(
        '← Arid / Sparse vegetation',
        xy=(0.02, -0.03),
        xycoords='axes fraction',
        ha='left',
        va='top',
        fontsize=12
    )

    ax.annotate(
        'Humid / Dense vegetation →',
        xy=(0.98, -0.03),
        xycoords='axes fraction',
        ha='right',
        va='top',
        fontsize=12
    )

    ax.annotate(
        'Thin soils; \ngentle slopes;\nlow soil conductivity ↑',
        xy=(-0.04, 0.85),
        xycoords='axes fraction',
        ha='right',
        va='bottom',
        fontsize=12
    )

    ax.annotate(
        'Deep soils; \nsteep slopes;\nhigh soil conductivity ↓',
        xy=(-0.04, 0.15),
        xycoords='axes fraction',
        ha='right',
        va='top',
        fontsize=12
    )

    ax.set_title(f'Dunne Diagram Clusters (k={n_clusters})', fontsize=14)

    for i, (x, y, dominants) in enumerate(
        zip(x_pos, y_pos, group_dominants), 1
    ):
        dom_text = "\n".join(
            [f"{k}: {v * 100:.1f}%" for k, v in dominants.items()]
        )

        text = (
            f"Group {i}\n"
            f"({group_size[i-1]} catchments)\n"
            f"{dom_text}"
        )

        ax.text(
            x,
            y,
            text,
            ha='center',
            va='center',
            fontsize=12,
            bbox=dict(
                boxstyle='round,pad=0.5',
                fc='lightyellow',
                ec='black',
                lw=1
            )
        )

    plt.tight_layout()
    plt.show()


# =========================
# Run clustering & plotting
# =========================
for n_clusters in range(3, 9):

    clusters = fcluster(Z, t=n_clusters, criterion='maxclust')
    full_df['cluster'] = clusters

    group_means = []
    group_dominants = []
    group_size = []

    print(f"\n=== Number of clusters: {n_clusters} ===")

    for c in range(1, n_clusters + 1):

        group = full_df[full_df['cluster'] == c]

        means = group[features].mean().to_dict()

        dominant = {
            'Ri': group['ri_ratio'].mean(),
            'Rs': group['rs_ratio'].mean(),
            'Qb': group['qb_ratio'].mean()
        }

        print(f"\nCluster {c}:")
        print("Number of basins:", len(group))
        print(
            "Mean component ratios:",
            {k: f"{v:.2f}" for k, v in dominant.items()}
        )
        print("Raw x, y:", compute_dunne_coords(means))

        group_means.append(means)
        group_dominants.append(dominant)
        group_size.append(len(group))

    plot_dunne_layout(
        group_means,
        group_dominants,
        n_clusters,
        group_size
    )
```
